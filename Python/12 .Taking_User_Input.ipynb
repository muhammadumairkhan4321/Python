# **Taking Input**

```python
# Taking user input example
# The input() function accepts data from the user.

Your_Name = input("Enter Your Name: ")
print("Hello,", Your_Name)
```

```python
# Taking user input example
# The input() function accepts data from the user.

Your_Age =input("Your Age :")
print("Age :",Your_Age)
```

```python
# Taking user input example
# The input() function accepts data from the user.

a = input("Enter Any Number")
b = input("Enter Another Name")

Sum = a + b

print("First No :",a)
print("Second No :",b)
print(Sum)
```

```python
# Taking user input example
# The input() function accepts data from the user.

a = int(input("Enter Any Number"))
b = int(input("Enter Another Name"))

Sum = a + b

print("First No :",a)
print("Second No :",b)
print(Sum)
```
