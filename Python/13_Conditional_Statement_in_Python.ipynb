# Conditional Statements in Python

## Introduction
Conditional statements allow programs to make decisions.

Topics:
- if
- if...else
- if...elif...else
- User Input
- lower() for case-insensitive comparison


```python
# Example of a simple if statement

Age = 21

if Age >=18:
    print("You Can vote")
```

```python
# Example of a simple if statement

Age = 21

if Age >=18:
    print("You Can vote")
    print("You Can Drive")
```

```python
# Example of if...else conditional statement

Age = 14

if Age >=18:
    print("You Can vote")
    print("You Can Drive")
else:
    print("You Can't vote")
    print("You Can't Drive")    
```

```python
# Take input from the user

Age = int(input("Enter Your age:"))

print("Your Age: ",Age)

if Age >=18:
    print("You Can vote")
    print("You Can Drive")
else:
    print("You Can't vote")
    print("You Can't Drive")    
```

```python
# Take input from the user

color = input("enter Color:")

print("Traffic Color: ",color)

if color == "red":
    print("Stop")
elif color == "green":
    print("Go")
elif color == "yellow":
    print("Look")    
```

```python
# Take input from the user

color = input("enter Color:")

print("Traffic Color: ",color)

if color == "red":
    print("Stop")
elif color == "green":
    print("Go")
elif color == "yellow":
    print("Look")    
```

```python
# Take input from the user

color = input("enter Color:")

print("Traffic Color: ",color)

if color == "red":
    print("Stop")
elif color == "green":
    print("Go")
elif color == "yellow":
    print("Look")    
else:
    print("Wrong color for Traffic light")   
```
