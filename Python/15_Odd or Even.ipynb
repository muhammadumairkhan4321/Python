## **Odd or Even**

```python
# Get input from the user and convert it to an integer
num = int(input("Please enter any number: "))

# Check if the number is completely divisible by 2 using the modulus operator (%)
if num % 2 == 0:
    # Print EVEN if the remainder is 0
    print(f"{num} is EVEN")
else:
    # Print ODD if the remainder is not 0
    print(f"{num} is ODD")
```
