````python
# Nested If Statement Example
# This code demonstrates how an if statement
# can be placed inside another if statement.

age = int(input("Enter your age: "))
has_id = input("Do you have an ID? (yes/no): ")

if age >= 18:
    if has_id == "yes":
        print("You are eligible to enter.")
    else:
        print("You need to show your ID.")
else:
    print("You are not eligible to enter.")
````

# Nested If Statement in Python

This file demonstrates the concept of a **Nested If Statement** in Python.

## What is a Nested If?

A Nested If is an `if` statement placed inside another `if` statement.

### Basic Syntax

```python
if condition:
    if another_condition:
        # Code to execute
    else:
        # Code to execute
else:
    # Code to execute

````python
age = int(input("Enter your age: "))
has_id = input("Do you have an ID? (yes/no): ")

if age >= 18:
    if has_id == "yes":
        print("You are eligible to enter.")
    else:
        print("You need to show your ID.")
else:
    print("You are not eligible to enter.")
````
