# Match Case in Python

This example demonstrates how to use Python's `match-case` statement for decision making.

## Example

The program asks the user to enter a traffic-light color and displays the appropriate action.

- `red` → `stop`
- `yellow` → `look`
- `green` → `go`
- Any other value → `invalid color`

## Concept

The `_` case works as the default/wildcard case when none of the specified patterns match.


```python
# Ask the user to enter a traffic-light color.
color = input("Enter a color: ")

# Match the user's input with the correct traffic-light action.
match color:
    case "red":
        print("stop")
    case "yellow":
        print("look")
    case "green":
        print("go")
    case _:
        # This default case runs when the color does not match.
        print("invalid color")
```
