# Infinite Loop and Finite Loop in Python

This notebook demonstrates the difference between an **infinite loop** and a **finite loop** in Python, including how to stop an infinite loop in Jupyter Notebook.

## Important
The infinite-loop example below is intentionally designed to run continuously. In Jupyter Notebook, stop it by using **Kernel → Interrupt Kernel** (the exact menu wording may vary by Jupyter version).

There is no need to use the old `I, I` shortcut in this example.

```python
# Infinite loop
# This loop continues until you interrupt the kernel.
while True:
    print("This is an infinite loop. Interrupt the kernel to stop it.")
```

## Finite Loop

A finite loop runs for a limited number of iterations and then stops automatically.

```python
# Finite loop
count = 1

while count <= 5:
    print("Hello World")
    count += 1
```

```python
# Finite loop with a counter
count = 1  # Counter variable

while count <= 10:
    print("Hello World", count)
    count += 1

print("The loop has ended.", count)
```
