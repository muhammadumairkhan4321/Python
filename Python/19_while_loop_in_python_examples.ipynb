# While Loop in Python

A `while` loop repeats a block of code as long as its condition is `True`.

## Examples
- Forward counting: `1` to `10`
- Reverse counting: `10` to `1`


```python
# Forward Counting
i = 1

while i <= 10:
    print(i)
    i += 1

print("Done")
```

```python
# Reverse Counting
i = 10

while i >= 1:
    print(i)
    i -= 1

print("Done")
```
