# Multiplication Table of a Number

This example uses a `while` loop to print the multiplication table of a number from 1 to 10.

### How it works
- The user enters a number.
- `i` starts from `1`.
- The loop continues while `i <= 10`.
- `i += 1` increases the counter after each iteration.


```python
# Ask the user to enter a number.
number = int(input("Enter any number: "))
print(number, "multiplication table is:")

# Start the counter at 1.
i = 1

# Print the multiplication table from 1 to 10.
while i <= 10:
    print(number * i)
    i += 1
```
