# Python `break` and `continue`

This notebook demonstrates how Python’s `break` and `continue` statements control the flow of `while` loops.

- `break` immediately stops the nearest loop.
- `continue` skips the rest of the current iteration and moves to the next one.

Each example below includes comments explaining the important lines.


## 1. Stop a loop with `break`

This loop starts at `1` and prints values until it reaches the first number divisible by `6`. At that point, `break` ends the loop, so `6` is not printed.


```python
# Start the counter at 1.
i = 1

# Keep looping while i is less than or equal to 10.
while i <= 10:
    # Stop the loop before printing the first multiple of 6.
    if i % 6 == 0:
        break

    # Print values that occur before the break condition.
    print(i)
    i += 1  # Move to the next number.

# This runs after the loop has ended.
print("outside loop now .....")
```

## 2. Use `break` in a descending loop

This example counts down from `10`. The loop stops as soon as it reaches a number divisible by `7`, which is `7`.


```python
# Start at 10 so the loop counts downward.
i = 10

# Continue until the counter goes below 1.
while i >= 1:
    # End the loop when the first multiple of 7 is reached.
    if i % 7 == 0:
        break

    print(i)
    i -= 1  # Decrease the counter for the next iteration.

print("outside loop now....")
```

## 3. Skip multiples of `3` with `continue`

When the current value is divisible by `3`, the counter is increased first and `continue` skips the `print()` statement. This prevents an infinite loop and ensures multiples of `3` are not displayed.


```python
# Start at the first number in the range.
i = 1

while i <= 10:
    # Skip values that are divisible by 3.
    if i % 3 == 0:
        i += 1  # Update i before continue to avoid an infinite loop.
        continue

    # Only non-multiples of 3 reach this line.
    print(i)
    i += 1

print("outside loop now .....")
```

## 4. Print odd numbers by stepping by `2`

Starting from `1`, adding `2` after every iteration naturally produces only odd numbers up to `10`.


```python
# Start with the first odd number.
i = 1

while i <= 10:
    print(i)
    i += 2  # Adding 2 keeps every value odd.

print("done")
```

## 5. Print odd numbers using `continue`

This version checks every number from `1` to `10`. Even numbers are skipped with `continue`, while odd numbers are printed.


```python
# Start at 1 and examine every number through 10.
i = 1

while i <= 10:
    # Skip even numbers.
    if i % 2 == 0:
        i += 1  # Update before continue so the loop progresses.
        continue

    # This line is reached only for odd numbers.
    print(i)
    i += 1

print("done")
```

## Key takeaway

Use `break` when you need to leave a loop completely. Use `continue` when you want the loop to continue but skip one particular iteration.
