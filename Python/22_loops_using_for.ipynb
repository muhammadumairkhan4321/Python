# Python For Loop Practice

This notebook contains beginner-friendly examples of Python `for` loops.

```python
# Print numbers from 1 to 10

for number in range(1, 11):
    print(number)
```

```python
# range(start, stop)

for number in range(1, 6):
    print(number)
```

```python
# Print even numbers from 2 to 20

for number in range(2, 21, 2):
    print(number)
```

```python
# Print odd numbers from 1 to 20

for number in range(1, 21, 2):
    print(number)
```

```python
# Print numbers from 1 to 10
# Stop the loop when i becomes 6

for i in range(1, 11):
    
    if i == 6:
        break
    
    print(i)
```

```python
# Print numbers from 1 to 10
# Skip number 5

for i in range(1, 11):
    
    if i == 5:
        continue
    
    print(i)
```

```python
# Print odd numbers using continue

for i in range(1, 11):
    
    if i % 2 == 0:
        continue
    
    print(i)
```

```python
word = "artificial intelligence"

# Count the number of i's

count = 0

for ch in word:
    if ch == 'i':
        count += 1

print("count of i =", count)
```
