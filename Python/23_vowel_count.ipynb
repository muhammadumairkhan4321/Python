# Python Vowel Count Using `for` Loop

This notebook counts lowercase vowels (`a`, `e`, `i`, `o`, and `u`) in a word using a Python `for` loop.


## How it works

- The loop checks each character in the text one by one.
- If the character is a vowel, `count` increases by `1`.
- The final result is printed after the loop finishes.


```python
# Store the text in a variable.
word = "artificial intelligence"

# Start the vowel counter at zero.
count = 0

# Check one character at a time in the word.
for char in word:
    # Count the character if it is a lowercase vowel.
    if char == "a" or char == "e" or char == "i" or char == "o" or char == "u":
        count += 1

# Display the total number of vowels.
print("Count of vowels =", count)
```
