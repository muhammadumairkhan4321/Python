# Python `range()` Function

This notebook demonstrates how to use Python’s `range()` function with a `for` loop. The `range()` function generates a sequence of numbers for iteration.


## 1. `range(stop)`

When `range()` receives one value, Python starts at `0` and stops before that value. Therefore, `range(10)` produces values from `0` through `9`.


```python
# range(10) starts at 0 and stops before 10.
for i in range(10):
    # Print the current number in the sequence.
    print(i)
```

## 2. `range(start, stop)`

When two values are provided, Python starts at the first value and stops before the second value. Therefore, `range(5, 10)` produces values from `5` through `9`.


```python
# Start at 5 and stop before 10.
for i in range(5, 10):
    # Print each number from 5 through 9.
    print(i)
```

## 3. `range(start, stop, step)`

The third value is the step size. Here, Python starts at `1`, adds `2` each time, and stops before `10`; therefore, it prints odd numbers from `1` through `9`.


```python
# Start at 1, stop before 10, and increase by 2 each time.
for i in range(1, 10, 2):
    # Print the current odd number.
    print(i)
```

## Key Takeaway

`range(stop)`, `range(start, stop)`, and `range(start, stop, step)` are useful for controlling where a `for` loop begins, ends, and how it advances.
