# Sum of the First N Numbers

This notebook tries to add the numbers from 1 to 5 using a running total (accumulator) variable.

**Idea:** start with `sum = 0`, then add each number to it, one at a time.

````python
# Goal: calculate the sum of the numbers 1 to 5 (1+2+3+4+5 = 15).
# `sum` starts at 0 and is meant to collect the running total.
# NOTE: the line `if i in range(1,6):` is a membership test, not a loop.
# Since `i` is never defined, running this cell raises: NameError: name 'i' is not defined.
# The intended code is a for loop: `for i in range(1,6):`  (left unchanged here on purpose).
# With the loop, the expected output would be: sum : 15
# Also note: the name `sum` shadows Python's built-in sum() function.
sum = 0

for i in range(1, 6):   # i takes the values 1, 2, 3, 4, 5
    sum += i

print("sum :", sum)     # sum : 15
````

### Notes
* The code as written **does not run**: `if i in range(1,6)` checks whether `i` is in the range, but `i` has not been defined, so Python raises a `NameError`.
* To visit every number from 1 to 5, a `for` loop is needed:
```python
sum = 0
for i in range(1, 6):   # i = 1, 2, 3, 4, 5
    sum += i
print("sum :", sum)    # sum : 15
```
* `range(1, 6)` includes 1 but excludes 6.
* Try to avoid naming a variable `sum`, since it hides the built-in `sum()` function.
