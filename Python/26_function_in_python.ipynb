# Functions in Python: Introduction

This notebook introduces **functions**, one of the most important building blocks in Python. You will learn how to define a function and how to call it.

## **Functions**
* block of statement that perform a specific task
* reuseble component of code  

### *Types*
* Definition
* Call

### Example 1: A simple function
The function below has one statement. Nothing runs when it is *defined*; it runs only when it is *called*.

```python
# Define a function named `hello` that takes no parameters.
# The indented line is the function body; it runs only when the function is called.
# Calling `hello()` afterwards executes the body and prints the message.
def hello(): # function definition
    print("hello")

hello() #function call     
```

### Example 2: A function with multiple statements
The same function name is defined again, this time with two `print` statements in its body. The new definition replaces the old one.

```python
# Redefine `hello` with two statements in its body.
# Python keeps only the latest definition, so this version replaces the previous one.
# Expected output: two lines, 'hello' and 'from python'.
def hello(): # function definition
    print("hello")
    print("from python")

hello() #function call     
```

### Key takeaways
* A function is defined with the `def` keyword, followed by a name, parentheses and a colon.
* The function body must be indented.
* A function does nothing until it is called using its name followed by `()`.
* Defining a function with an existing name overwrites the old one.
