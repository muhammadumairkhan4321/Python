# Functions: Practice Examples

This notebook practises functions with **parameters**, **return values** and **default parameters**, and shows a common mistake that raises a `SyntaxError`.

### Example 1: Average of three numbers
The function receives three numbers, adds them and returns the average.

```python
# Function that calculates the average of three numbers.
# `sum` here is a local variable holding a+b+c (note: it shadows the built-in sum() inside the function).
# `return sum/3` sends the average back to the caller.
# Expected output: Average 2.0  ((1+2+3)/3)
# calculate average
def cal_avg(a,b,c):
    sum = a+b+c
    return sum/3

print("Average",cal_avg(1,2,3))
```

```python
# Same function, called with different arguments.
# Expected output: Average 5.666666666666667  ((5+9+3)/3)
# calculate average
def cal_avg(a,b,c):
    sum = a+b+c
    return sum/3

print("Average",cal_avg(5,9,3))
```

### Default parameters
A parameter can have a **default value** that is used when the caller does not supply that argument.

```python
# `b=6` is a default parameter: if `b` is not passed, 6 is used.
# Here both arguments are passed (5 and 6), so the default is overridden.
# Note: defining `sum` here shadows Python's built-in sum() function.
# default parameters

def sum(a,b=6):
    return a+b

print(sum(5,6)) # print 11
```

```python
# Only `a` is passed (5), so `b` takes its default value of 1.
# Result: 5 + 1 = 6
# default parameters

def sum(a,b=1):
    return a+b

print(sum(5)) # print 6
```

### Common mistake: non-default parameter after a default one
The next cell is **expected to fail**. It is kept to show that Python does not allow this parameter order.

```python
# WRONG ORDER: a parameter with a default (a=1) cannot come before a parameter without one (b).
# Python raises: SyntaxError: parameter without a default follows parameter with a default.
# Fix: put non-default parameters first, e.g. def sum(b, a=1)  (this cell is left as-is on purpose).
# Because of the error, the code below never runs, so the '# print 6' comment does not apply.
# default parameters

def sum(a=1,b):
    return a+b

print(sum(5)) # print 6
```

### Key takeaways
* `return` sends a value back to the place where the function was called.
* Default parameters make some arguments optional.
* Parameters **with** defaults must come **after** parameters **without** defaults.
* Avoid naming your own functions or variables `sum`, since that hides Python's built-in `sum()`.
