# Types of Functions in Python

This short notebook lists the two main categories of functions in Python.

## **Functions**

### *Types*
* Bult-in e.g print(),input(),type(),range()
* user-defined 

### Explanation

**1. Built-in functions** are provided by Python and are always available, for example:
* `print()` displays output
* `input()` reads text typed by the user
* `type()` returns the type of a value
* `range()` generates a sequence of numbers

**2. User-defined functions** are created by the programmer with the `def` keyword (or `lambda`) to perform a custom task, as seen in the previous notebooks.

*(Note: "Bult-in" in the list above is a typo for "Built-in".)*
