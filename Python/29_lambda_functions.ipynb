# Lambda Functions in Python

A **lambda function** is a small, anonymous (nameless) function written in a single line.

**Syntax:** `lambda arguments: expression`

The expression is evaluated and its result is returned automatically (no `return` keyword needed).

## **Lambda Function**

### Example 1: Sum of two numbers

````python
# Create a lambda that takes two arguments (a, b) and returns their sum.
# It is stored in the variable `sum`, so it can be called like a normal function.
# Expected output: 9  (4 + 5)
sum = lambda a,b: a+b

print(sum(4,5)) #print 9 
````

### Example 2: Average of two numbers

````python
# Lambda that returns the average of two numbers.
# Expected output: 4.5  ((4 + 5) / 2)
avg = lambda a,b: (a+b)/2

print(avg(4,5)) #print 4.5 
````

* Lambda
* usage high order function

### Notes
* A lambda can contain only **one expression**; it cannot hold multiple statements.
* Lambdas are mostly used with **higher-order functions**, i.e. functions that take another function as an argument (such as `map()`, `filter()` and `sorted()`).

Example (for reference only):
```python
nums = [1, 2, 3]
squares = list(map(lambda x: x * x, nums))   # [1, 4, 9]
```
