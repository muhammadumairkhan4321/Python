# Factorial of a Number

The factorial of a number `n` (written `n!`) is the product of all whole numbers from 1 to `n`.

Examples: `5! = 1 x 2 x 3 x 4 x 5 = 120`. By definition, `0! = 1`.

## **Factorail**
* n! --> 1x2x3x4x5-----xn

### Example 1: n = 5

```python
# Function that returns the factorial of n using a for loop.
# `fact` starts at 1 and is multiplied by every number from 1 up to n.
# range(1, n+1) is used because the end value of range() is excluded.
# The user enters n with input(); int() converts the text into an integer.
# Sample run: n = 5 -> output 120
def calc_factorail(n):
    fact = 1
    for i in range(1,n+1):
        fact *= i

    return fact

n = int(input("enter n : "))
print("number :",n)
print(calc_factorail(n))        
```

### Example 2: n = 12

```python
# Same function, run again with a different input.
# Sample run: n = 12 -> output 479001600
def calc_factorail(n):
    fact = 1
    for i in range(1,n+1):
        fact *= i

    return fact

n = int(input("enter n : "))
print("number :",n)
print(calc_factorail(n))     
```

### Notes
* The loop multiplies `fact` by `i` on every iteration (`fact *= i`).
* For `n = 0` the loop does not run, so the function returns 1, which is correct.
* Negative numbers are not handled by this code (their factorial is undefined).
