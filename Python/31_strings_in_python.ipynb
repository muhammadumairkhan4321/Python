# Strings in Python

A **string** is a sequence of characters enclosed in quotes. This notebook covers length, concatenation, indexing, and looping over a string.

### String length
`len()` returns the number of characters in a string.

```python
# Store a string and count its characters with len().
# Expected output: characters: 6
word = "python"

print("characters:",len(word))
```

### Concatenation
Strings can be joined with the `+` operator.

```python
# Join two strings using +. No space is added automatically.
# Expected output: Muhammadumair
# concatenate

word1 = "Muhammad"
word2 = "umair"

print(word1 + word2)
```

```python
# Add a space string " " between the two words to separate them.
# Expected output: Muhammad umair
# concatenate

word1 = "Muhammad"
word2 = "umair"

print(word1 +" "+ word2)
```

```python
# Store the joined result in a variable (`sentance`) and print it later.
# Expected output: Muhammad umair
# concatenate

word1 = "Muhammad"
word2 = "umair"
sentance = word1 +" "+ word2

print(sentance)
```

### *indexing*

Each character has a position called an **index**, starting from **0**. For `"python"`: p=0, y=1, t=2, h=3, o=4, n=5.

```python
# Access a single character using its index. Index 2 is the third character.
# Expected output: t
word = "python"

print(word[2])
```

### Looping over a string
A `for` loop can visit every character one by one.

```python
# Loop through the string; `ch` takes each character in turn.
# Expected output: p, y, t, h, o, n (one per line)
word = "python"

for ch in word:
    print(ch)
```

### Strings are immutable
Once a string is created, its characters **cannot be changed in place**. Trying `word[0] = "P"` raises a `TypeError`. To get a modified string, create a new one, for example `"P" + word[1:]`.
