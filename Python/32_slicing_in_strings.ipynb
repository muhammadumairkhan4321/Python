# String Slicing in Python

Slicing extracts a **substring** using the syntax `string[start:end]`.
* `start` is included, `end` is **excluded**.
* If `start` is omitted it defaults to 0; if `end` is omitted it goes to the end of the string.
* Negative indices count from the end (`-1` is the last character).

## **Strings**
### *Slicing*

* substrings

### Basic slicing

```python
# word[2:6] returns characters at index 2, 3, 4, 5 (index 6 is excluded).
# Expected output: thon
word = "python"

print(word[2:6])
```

### Slicing a longer sentence
First, count the characters in the sentence so we know the index positions.

```python
# len() counts every character, including spaces.
# Expected output: Characters : 24
word = "i study from ApnaCollege"

print("Characters :",len(word))
```

```python
# In this sentence 'A' of ApnaCollege is at index 13, so word[13:17] takes indices 13-16.
# Expected output: Apna
word = "i study from ApnaCollege"

print(word[13:17])
```

```python
# Omitting the end index slices from index 13 to the end of the string.
# Expected output: ApnaCollege .
word = "i study from ApnaCollege ."

print(word[13:])
```

```python
# word[:] with no start or end returns a copy of the whole string.
# Expected output: i study from ApnaCollege .
word = "i study from ApnaCollege ."

print(word[:])
```

```python
# A negative index counts from the end; -1 is the last character.
# Expected output: n
word1= "python"

print(word1[-1])
```

```python
# Negative slice. The string has 6 characters, so -7 is out of range and is treated as the start (0).
# -4 points to index 2 (excluded), so the result is the first two characters.
# Slicing never raises an error for out-of-range values. Expected output: py
word1= "python"

print(word1[-7:-4])
```

### Quick reference
| Expression | Meaning |
|---|---|
| `s[a:b]` | characters from index `a` up to (not including) `b` |
| `s[a:]` | from index `a` to the end |
| `s[:b]` | from the start up to (not including) `b` |
| `s[:]` | the whole string |
| `s[-1]` | the last character |
