# String Formatting in Python

## **Formatting**
--> dynamic strings
    (var & vals)
1. Format()
2. f-strings    

**Overview:** String formatting lets you build *dynamic strings* by inserting variables and values into a text template.

Two common approaches:
1. `format()` method: `"text {}".format(value)`
2. f-strings: `f"text {value}"` (available since Python 3.6)

### Using `format()` with one placeholder
Each `{}` in the string is replaced by a value passed to `.format()`.

````python
# Build a message with a variable using str.format().
# `{}` is replaced by the value of `sum`. Expected output: sum is 15
a = 10
b = 5
sum = a + b

# normal formatting
print("sum is {}".format(sum))
````

### Placeholder with a literal value

````python
# The value can also be given directly instead of using a variable.
# Expected output: Language is python
print("Language is {}".format("python"))
````

### Multiple placeholders
Values are inserted into the `{}` slots in the same order they are given.

````python
# Three placeholders are filled by a, b and sum in order.
# Expected output: sum of 10 & 5 is 15
a = 10
b = 5
sum = a + b

# normal formatting
print("sum of {} & {} is {}".format(a,b,sum))
````

### Index-based formatting
Numbers inside the braces (`{0}`, `{1}`, `{2}`) select values by position. See the note in the cell below.

````python
# NOTE: This cell has not been executed yet (no output shown).
# It uses empty {} placeholders, which fill in order, just like the previous cell.
# A truly index-based version would use numbered placeholders: {0}, {1}, {2}.
#index based formatting
print("sum of {} & {} is {}".format(a,b,sum))
````

### f-strings (for reference)
An f-string puts the variable directly inside the braces:
```python
a = 10
b = 5
print(f"sum of {a} & {b} is {a + b}")   # sum of 10 & 5 is 15
```
f-strings are usually shorter and easier to read than `format()`.
