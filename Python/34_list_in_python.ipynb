# Lists in Python

A **list** is an ordered collection of values written inside square brackets `[]`, with items separated by commas. Lists are **mutable**, which means their items can be changed after the list is created.

## **Lists**

* mutable squence of values

### Creating a list and finding its length
`len()` returns the number of items in a list.

```python
# Create a list of four marks, then print the whole list and its length.
# Expected output: [45, 43, 39, 50] and 4
marks = [45,43,39,50]

print(marks)
print(len(marks))
```

### Accessing items by index
Indexing starts at **0**, so index 2 is the third item.

```python
# Access one item using its index. Index 2 is the third item (39).
# Expected output: index 2 value is : 39
marks = [45,43,39,50]

print("index 2 value is :",marks[2])
```

### Changing an item (mutability)
Unlike strings, list items can be replaced by assigning to an index.

```python
# Lists are mutable: replace the item at index 2 (39) with 44.
# Expected output: [45, 43, 44, 50]
marks = [45,43,39,50]
marks[2] = 44

print(marks)
```

### A list of strings

```python
# A list of student names (strings). len() counts the items.
# Expected output: the full list, then 6
students = ["Ali" , "Murad","Ihsan","Kashif","Abdul Aziz","Khan"]

print(students)
print(len(students))
```

### Slicing a list
Slicing works like string slicing: `list[start:end]` includes `start` and excludes `end`.

```python
# students[0:4] returns items at index 0, 1, 2, 3 (index 4 is excluded).
# Expected output: ['Ali', 'Murad', 'Ihsan', 'Kashif']
students = ["Ali" , "Murad","Ihsan","Kashif","Abdul Aziz","Khan"]

print(students[0:4])
```

```python
# Leaving out the start index is the same as starting at 0, so [:4] equals [0:4].
# Expected output: ['Ali', 'Murad', 'Ihsan', 'Kashif']
students = ["Ali" , "Murad","Ihsan","Kashif","Abdul Aziz","Khan"]

print(students[:4])
```

```python
# Leaving out the end index slices from index 4 to the end of the list.
# Expected output: ['Abdul Aziz', 'Khan']
students = ["Ali" , "Murad","Ihsan","Kashif","Abdul Aziz","Khan"]

print(students[4:])
```

### Lists can hold mixed data types
One list can contain integers, strings and other types together.

```python
# A list can mix data types: here numbers and the string 'abc'.
# Expected output: [89, 98, 70, 56, 86, 'abc', 67, 76]
marks = [89,98,70,56,86,"abc",67,76]

print(marks)
```

### Negative slicing
Negative indices count from the end of the list.

```python
# marks[-3:] takes the last three items of the list.
# Expected output: ['abc', 67, 76]
marks = [89,98,70,56,86,"abc",67,76]

print(marks)
print(marks[-3:])
```

### Key takeaways
* Lists are created with `[]` and keep their items in order.
* Indexing starts at 0; negative indices count from the end.
* Lists are **mutable**: `marks[2] = 44` changes the list in place.
* Slicing `list[start:end]` returns a new list and excludes `end`.
* A list can hold different data types together.
