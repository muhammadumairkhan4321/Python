# List Methods in Python

A **method** is a function that belongs to an object. A list is an object of the built-in `list` class, so it comes with ready-made methods such as `append()`, `insert()`, `sort()` and `reverse()`. They are called with dot notation: `list_name.method()`.

## **List**
### *Methods*

* Methods----> Functions
* list--->Classes--->Functions--->Methods

### `append()`: add an item at the end

```python
# append() adds one item at the end of the list.
# Expected output: [1, 2, 3, 4, 5]
numbers = [1,2,3,4]
numbers.append(5)

print(numbers)
```

### `insert()`: add an item at a chosen position
Syntax: `list.insert(index, value)`

```python
# insert(2, 5) puts the value 5 at index 2; existing items from index 2 move one place right.
# Expected output: [1, 2, 5, 3, 4]
numbers = [1,2,3,4]
numbers.insert(2,5)

print(numbers)
```

### `sort()`: arrange items in ascending order
This changes the original list in place.

```python
# Insert 10 at index 2, then sort() arranges the list in ascending order (in place).
# Expected output: [1, 2, 10, 3, 4] then [1, 2, 3, 4, 10]
numbers = [1,2,3,4]
numbers.insert(2,10)

print(numbers)

numbers.sort()
print(numbers)
```

### `sort(reverse=True)`: descending order

```python
# sort(reverse=True) arranges the list in descending order.
# Expected output: [1, 2, 10, 3, 4] then [10, 4, 3, 2, 1]
numbers = [1,2,3,4]
numbers.insert(2,10)

print(numbers)

numbers.sort(reverse=True)
print(numbers)
```

### `reverse()`: flip the current order
This only reverses the existing order; it does not sort.

```python
# Sort the list first, then reverse() flips the order of the sorted list.
# Expected output: [1, 2, 10, 3, 4], then [1, 2, 3, 4, 10], then [10, 4, 3, 2, 1]
numbers = [1,2,3,4]
numbers.insert(2,10)

print(numbers)

numbers.sort()
print(numbers)

numbers.reverse()
print(numbers)
```

### Summary
| Method | What it does |
|---|---|
| `append(x)` | adds `x` at the end |
| `insert(i, x)` | adds `x` at index `i` |
| `sort()` | sorts ascending (in place) |
| `sort(reverse=True)` | sorts descending (in place) |
| `reverse()` | reverses the current order (in place) |

**Note:** these methods change the original list and return `None`, so writing `numbers = numbers.sort()` would set `numbers` to `None`.
