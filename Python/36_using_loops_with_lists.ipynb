# Using Loops with Lists

A `for` loop can visit every item of a list one by one. This notebook shows how to print all items and how to use a loop to search for a value (linear search).

### Example 1: Print every item of a list

```python
# `val` takes each item of the list `num` in turn, and print() displays it.
# Expected output: the 9 numbers printed one per line, in the original order.
num = [1,2,3,4,6,7,5,8,0]

for val in num:
    print(val)
```

### Example 2: Linear search
**Linear search** checks the items one after another from the start until the target value is found. A counter variable keeps track of the current index.

```python
# Goal: find the index of the value x = 8 in the list.
# `idx` is a counter that follows the current position in the list (starts at 0).
# For each item: if it equals x, print its index and stop with break; otherwise increase idx by 1.
# Expected output: 8 found at index = 7
# Note: if x is not in the list, this code prints nothing.
# Linear search
num = [1,2,3,4,6,7,5,8,0]
x = 8
idx = 0

for val in num:
    if (val==x):
        print(f"{x} found at index = {idx}")
        break
    idx +=1
```

### Key takeaways
* `for val in list:` gives each item of the list, one at a time.
* `break` stops the loop as soon as the item is found, so no extra work is done.
* The counter `idx` is increased only when the item does not match.
* In a linear search the worst case checks every item in the list.
