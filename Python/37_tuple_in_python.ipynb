# Tuples in Python

A **tuple** is an ordered collection of values written inside round brackets `()`. Unlike a list, a tuple is **immutable**: its items cannot be changed after it is created.

### Curly braces create a set, not a tuple
The next two cells use `{}` even though the variable is named `tup`. Curly braces create a **set**, which is a different data type.

## **Tuples**
* immutable sequence of values

```python
# NOTE: {} creates a SET, not a tuple, despite the variable name `tup`.
# A set is unordered and keeps no duplicates, so the printed order may differ from the written order.
# Expected output: {1, 2, 3, 4, 5, 6, 7, 8, 'abc'}
tup = {1,2,3,4,"abc",5,6,7,8}

print(tup)
```

### Creating a real tuple
Use round brackets `()` and check the type with `type()`.

```python
# This cell is expected to fail (it is kept to show the error).
# `tup` is a set, and sets do not support item assignment or indexing.
# Error: TypeError: 'set' object does not support item assignment
# A real tuple would also fail here, with: 'tuple' object does not support item assignment.
tup = {1,2,3,4,"abc",5,6,7,8}
tup[4]=5
print(tup)
```

### Single-item tuples: the comma matters
Brackets alone do not make a tuple. A one-item tuple needs a trailing comma. The next three cells show this.

```python
# Round brackets create a tuple. A tuple can mix data types (numbers and a string).
# Expected output: <class 'tuple'>
tup = (1,2,3,4,5,"abc",6,7,8)
print(type(tup))
```

```python
# (1) is NOT a tuple: the brackets only group the value, so it stays an integer.
# Expected output: <class 'int'>
tup = (1)
print(type(tup))
```

```python
# ("khan") is NOT a tuple either; it is just a string in brackets.
# Expected output: <class 'str'>
tup =("khan")
print(type(tup))
```

```python
# A trailing comma makes a one-item tuple: ("khan",)
# Expected output: <class 'tuple'>
tup =("khan",)
print(type(tup))
```

### Slicing a tuple
Tuples support indexing and slicing just like lists and strings.

```python
# Slicing works like lists: tup[:3] returns items at index 0, 1, 2 as a new tuple.
# Expected output: (1, 2, 3)
tup = (1,2,3,4,5)

print(tup[:3])
```

### Key takeaways
* Tuples use `()` and are **immutable**; lists use `[]` and are mutable; sets use `{}`.
* A one-item tuple needs a trailing comma: `(5,)`, not `(5)`.
* Tuples support indexing and slicing, but not item assignment.
* Use a tuple for data that should not change.
