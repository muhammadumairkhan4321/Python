# Tuple Methods and Looping Over Tuples

This notebook shows how to loop over a tuple and how to use the tuple method `count()`. Tuples are immutable, so they have only two methods: `count()` and `index()`.

### Looping over a tuple
A `for` loop visits each item of a tuple, just like with a list.

````python
# Loop through the tuple; `val` takes each item in turn and print() displays it.
# Expected output: numbers 1 to 6, one per line.
tup = (1,2,3,4,5,6)

for val in tup:
    print(val)
````

### Sum of all items
A running total (`sum`) is updated inside the loop.

````python
# `sum` starts at 0 and each item of the tuple is added to it inside the loop.
# Expected output: sum of vals is 21  (1+2+3+4+5+6)
# Note: the name `sum` shadows Python's built-in sum() function.
tup = (1,2,3,4,5,6)

sum = 0

for val in tup:
    sum += val
print(f"sum of vals is {sum}")    
````

Tuples are immutable, so methods that change a tuple (like `append()` or `sort()`) do not exist.

### *Methods*

````python
# count(2) returns how many times the value 2 appears in the tuple.
# Expected output: 4
tup = (1,2,3,4,5,2,5,1,2,1,1,2,3,4)

print(tup.count(2))
````

### Summary
| Method | What it does |
|---|---|
| `tuple.count(x)` | returns how many times `x` appears |
| `tuple.index(x)` | returns the index of the first occurrence of `x` |

Example (for reference only):
```python
tup = (10, 20, 30, 20)
print(tup.index(20))   # 1
```
